### 채팅모델 사용해보기

```python
# 코드실행에 필요한 라이브러리를 상단위치에 선언.
from langchain_openai import ChatOpenAI
```

```python
# 쳇모델 객체 생성
llm = ChatOpenAI(
    temperature=0.1,  # 창의성 (0.0 ~ 2.0)
    model_name="gpt-4o-mini",  # 모델명   
)

# 질문
question = "대한민국의 수도는 어디인가요?"

# 응답
response = llm.invoke(question)
```

```python
print(f"[답변]: {response}")
```

```python
# 사용된 전체토큰 : 24
response.response_metadata["token_usage"]["total_tokens"]
```

### logprob 로그 프로버블리티 활성화
- 모델이 단어(토큰)를 생성할 때 "이 단어를 얼마나 확신을 가지고 선택했는지"에 대한 확률값을 얻을 수 있다.

##### 모델의 답변 확신도(Confidence level) :0에 가까울수록 정답에 가까운 확신을 가진다.

```python
# 객체 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1,  # 창의성 (0.0 ~ 2.0)
    max_tokens=2048,  # 최대 토큰수
    model_name="gpt-4o-mini",  # 모델명 
).bind(logprobs=True)
```

```python
# 질의내용
question = "대한민국의 수도는 어디인가요?"

# 질의
response = llm_with_logprob.invoke(question)
```

```python
# 결과 출력
response.response_metadata
```

```python
# "서울" : 0에 가까운 확신답변
response.response_metadata["logprobs"]
```

### Steam 실시간 출력 방식

```python
# 답변을, 한토큰씩 받아와서 화면에 실시간으로 뿌려줄때.
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")
```

```python
# flush=False: 결과가 버퍼에 저장되었다가, 한번에 출력.
# flush=True : 결과가 생성되는대로 바로바로 출력.
for token in answer:
    print(token.content, end="", flush=True)
```

```python
# 스트리밍 답변결과 변수에 저장
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")

final_answer = ""
for token in answer:
    final_answer += token.content
```

```python
# 저장된 결과 출력
final_answer
```
