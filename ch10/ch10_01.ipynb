```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
with open("../data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()
```

```python
# 500자 출력
print(file[:500])
```

- 텍스트 분할기준 기본값 : "\n\n"

```python
# page 307
text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=210,
    chunk_overlap=0,
    length_function=len
)
```

```python
# [file] : 리스트에 담아서 사용.
texts = text_splitter.create_documents([file])
print(len(texts[0].page_content))
print(texts[0])
```

```python
# 원본문서 file {"document":1} : 32개의 청크로 나눈 원본
# 원본문서 file {"document":2} : 32개의 청크로 나눈 원본
# 똑같은 원본 2개로, 문서에 메타파일 셋팅을 보여줌.
metadatas=[
    {"document":1},
    {"document":2}
]

documents = text_splitter.create_documents([
    file,
    file
],
metadatas=metadatas
)

print(documents[0])
print("-----")
print(len(documents))
print("-----")
print(documents[32])
```
