### Prompt의 부분 변수 활용하기

```python
from  datetime import datetime

datetime.now().strftime("%B %d")
```

```python
def get_today():
    return datetime.now().strftime("%B %d")
```

```python
prompt = PromptTemplate(
    template="오늘의 날짜는 {today}입니다. 오늘이 생일인 유명인 {n}명을 나열해 주세요. 생년월일을 표기해 주세요",
    input_variables=["n"],
    partial_variables={
        "today":get_today
    }
)
```

```python
prompt.format(n=3)
```

```python
chain = prompt | llm

chain.invoke(3).content
```

```python
print(chain.invoke(3).content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_promt = ChatPromptTemplate.from_template("{country}의 수도는 어디인가요?")
chat_promt
```

```python
chat_promt.format(country="인도")
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate(
    [
        ("system","당신은 친절한 AI 어시스턴트입니다. 당신의 이름은 {name}입니다."),
        ("human", "반가워요!"),
        ("ai", "안녕하세요! 무엇을 도와드릴까요?"),
        ("human", "{user_input}")
    ]
)
chat_template
```

```python
messages = chat_template.format_messages(
    name="선경쌤", user_input="당신의 이름은 무엇입니까?"
)
messages
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()
llm.invoke(messages).content
```

```python
chain = chat_template | llm

chain.invoke({"name": "skc", "user_input":"당신의 이름은 무엇입니까?"}).content
```
