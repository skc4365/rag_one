```python
from langchain_community.document_loaders.csv_loader import CSVLoader

FILE_PATH = "../data/titanic.csv"
```

```python
loader = CSVLoader(FILE_PATH)

docs = loader.load()

print(len(docs))
print("----------")
print(docs[0].metadata)
```

```python
print("1번 승객 정보 : \n", docs[0].page_content)
print("----------")
print("2번 승객 정보 : \n", docs[1].page_content)
```

```python
loader = CSVLoader(
    file_path=FILE_PATH,
    csv_args={
        "delimiter": ",",  # 구분자
        "quotechar": '"',  # 인용 부호 문자
        "fieldnames": [
            "승객 ID",
            "생존 (1: Survived, 0: Died)",
            "Passenger Class",
            "이름",
            "성별",
            "나이",
            "Number of Siblings/Spouses Aboard",
            "Number of Parents/Children Aboard",
            "Ticket Number",
            "Fare",
            "Cabin",
            "Port of Embarkation",
        ],  # 필드 이름
    }
)
```

```python
docs = loader.load()
print(docs[1].page_content)
```

```python
row = docs[1].page_content.split("\n")
row_str = "<row>"
for element in row:
    splitted_element = element.split(":")
    value = splitted_element[-1]
    col = ":".join(splitted_element[:-1])
    row_str += f"<{col}>{value.strip()}</{col}>"
row_str += "</row>"
print(row_str)
```

```python
for doc in docs[1:]:
    row = doc.page_content.split("\n")
    row_str = "<row>"
    for element in row:
        splitted_element = element.split(":")
        value = splitted_element[-1]
        col = ":".join(splitted_element[:-1])
        row_str += f"<{col}>{value.strip()}</{col}>"
    row_str += "</row>"
    print(row_str)
```
